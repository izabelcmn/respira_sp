```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_cetesb = pd.read_csv("../../data/dados_pm25_estacao8_2013_2019.csv")
df_cetesb.head()
```

```python
df_atmosfera = pd.read_csv("../../data/openmeteo_historico_2013_2019.csv")
df_atmosfera.head()
```

```python
df_cetesb["time"] = (
    pd.to_datetime(df_cetesb["time"])
    .dt.tz_localize("Etc/GMT+3")
    .dt.tz_convert("UTC")
)

df_atmosfera["time"] = (
    pd.to_datetime(df_atmosfera["time"], utc=True)
)
```

```python
df = pd.merge(
    df_cetesb[["time", "MP2.5"]],
    df_atmosfera[["time", "temperature_2m", "relative_humidity_2m", 'precipitation', 'wind_speed_10m']],
    on="time",
    how="inner"
)

df = df.set_index("time")
```

### Analisando gaps do target

```python
serie = df["MP2.5"]

is_gap = serie.isna()

grupos = (is_gap != is_gap.shift()).cumsum()

gaps = (
    df[is_gap]
    .groupby(grupos[is_gap])
    .apply(lambda x: pd.Series({
        "inicio": x.index.min(),
        "fim": x.index.max(),
        "duracao_h": len(x)
    }))
)

gaps
```

```python
gaps.shape
```

```python
gaps_pequenos = gaps[gaps["duracao_h"] <= 12]
gaps_grandes = gaps[gaps["duracao_h"] > 12]

print("Gaps <= 12h:", len(gaps_pequenos))
print("Gaps > 12h:", len(gaps_grandes))
```

```python
gaps["duracao_h"].max()
```

```python
gaps.sort_values("duracao_h", ascending=False).head(10)
```

```python
gaps["ano_inicio"] = gaps["inicio"].dt.year
gaps["mes_inicio"] = gaps["inicio"].dt.month
```

```python
gaps.head()
```

```python
gaps.groupby("ano_inicio")["duracao_h"].agg(["count", "mean", "max", "sum"])
```

### Estratégia 2 —  com interpolação <= 12h X Todas as features (2013-2019)

```python
df_estrategia2 = df.copy()
```

```python
df_estrategia2 = df.copy()

# Série interpolada usando o contexto completo antes/depois dos gaps
mp25_interpolado = df["MP2.5"].interpolate(method="time", limit_area="inside")

# Seleciona apenas gaps pequenos
gaps_pequenos = gaps[gaps["duracao_h"] <= 12]

# Preenche somente os intervalos dos gaps pequenos
for _, gap in gaps_pequenos.iterrows():
    inicio = gap["inicio"]
    fim = gap["fim"]

    df_estrategia2.loc[inicio:fim, "MP2.5"] = mp25_interpolado.loc[inicio:fim]
```

```python
print("NaNs antes:", df["MP2.5"].isna().sum())
print("NaNs depois:", df_estrategia2["MP2.5"].isna().sum())
```

```python
# série interpolada
serie_interp = df_estrategia2["MP2.5"]

# identificar NaNs remanescentes
is_gap = serie_interp.isna()

# grupos contínuos de NaN
grupos = (is_gap != is_gap.shift()).cumsum()

# tabela de gaps após interpolação
gaps_interp = (
    df_estrategia2[is_gap]
    .groupby(grupos[is_gap])
    .apply(lambda x: pd.Series({
        "inicio": x.index.min(),
        "fim": x.index.max(),
        "duracao_h": len(x)
    }))
)

print("NaNs remanescentes:", df_estrategia2["MP2.5"].isna().sum())
print("Soma dos gaps:", gaps_interp["duracao_h"].sum())
print(gaps_interp["duracao_h"].describe())
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    gaps_interp["duracao_h"],
    bins=np.logspace(
        np.log10(1),
        np.log10(gaps_interp["duracao_h"].max()),
        30
    )
)

plt.xscale("log")
plt.xlabel("Duração do gap (h)")
plt.ylabel("Quantidade")
plt.title("Gaps remanescentes após interpolação")
plt.grid(alpha=0.3)

plt.show()
```

```python
df_estrategia2.to_csv(
    "../../data/dados_2013_2019_interp_12h_all_features.csv",
    index=True
)
```

### Estratégia 3 —  com interpolação <= 12h + imputação sazonal nos gaps grandes X Todas as features (2013-2019)

```python
df_estrategia3 = df.copy()
```

```python
mp25_interpolado = df["MP2.5"].interpolate(method="time", limit_area="inside")


gaps_pequenos = gaps[gaps["duracao_h"] <= 12]


for _, gap in gaps_pequenos.iterrows():
    inicio = gap["inicio"]
    fim = gap["fim"]

    df_estrategia3.loc[inicio:fim, "MP2.5"] = mp25_interpolado.loc[inicio:fim]
```

```python
# cria variáveis temporais
df_estrategia3["mes"] = df_estrategia3.index.month
df_estrategia3["hora"] = df_estrategia3.index.hour
df_estrategia3["dia_semana"] = df_estrategia3.index.dayofweek

# flag: o que ainda está NaN depois da interpolação <=12h
df_estrategia3["MP2.5_imputado_media"] = df_estrategia3["MP2.5"].isna().astype(int)
```

```python
# média do mesmo mês + dia da semana + hora
media_mdh = (
    df_estrategia3
    .groupby(["mes", "dia_semana", "hora"])["MP2.5"]
    .transform("mean")
)
media_mdh
```

```python
# preenche gaps maiores com essa média
df_estrategia3["MP2.5"] = df_estrategia3["MP2.5"].fillna(media_mdh)
```

```python
df_estrategia3.head()
```

```python
df_estrategia3.to_csv(
    "../../data/pm25_2013_2019_interp_12h_media_sazonal_all_features.csv",
    index=True
)
```

```python
fig, ax = plt.subplots(figsize=(15,5))



ax.plot(
    df_estrategia3.index,
    df_estrategia3["MP2.5"],
    label="Estratégia 3",
    alpha=0.7
)

ax.plot(
    df.index,
    df["MP2.5"],
    label="Original",
    alpha=0.7
)

ax.set_ylabel("PM2.5 (µg/m³)")
ax.set_title("Comparação da série original e série preenchida")
ax.legend()

plt.show()
```
